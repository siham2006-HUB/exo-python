```python
import os
import seaborn as sns

titanic = sns.load_dataset("titanic")
df = titanic.copy()
df.shape
```

```python
nb_age_manquants = df["age"].isna().sum()
mediane_age = df.groupby(["class", "sex"], observed=True)["age"].transform("median")
df["age"] = df["age"].fillna(mediane_age)

nb_port_manquants = df["embark_town"].isna().sum()
port_le_plus_frequent = df["embark_town"].mode()[0]
df["embark_town"] = df["embark_town"].fillna(port_le_plus_frequent)

df = df.drop(columns=["deck"])
print(nb_age_manquants, nb_port_manquants, port_le_plus_frequent)
```

```python
nb_doublons = titanic.duplicated().sum()
print("Doublons :", nb_doublons)
```

```python
billets_gratuits = df[df["fare"] == 0]
billets_tres_chers = df[df["fare"] > 500]
print("Billets à 0 :", len(billets_gratuits), "| Billets > 500 :", len(billets_tres_chers))
```

```python
os.makedirs("data", exist_ok=True)
df.to_csv("data/titanic_propre.csv", index=False)
print(df.shape, "- manquants :", df.isna().sum().sum())
```
