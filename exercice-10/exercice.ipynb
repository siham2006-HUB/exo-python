```python
import pandas as pd

titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")
ports = pd.DataFrame({"embark_town": ["Southampton", "Cherbourg", "Queenstown"],
                      "pays": ["Angleterre", "France", "Irlande"]})

titanic_pays = titanic.merge(ports, on="embark_town", how="left")
len(titanic_pays)
```

```python
titanic_pays.groupby("pays")["survived"].agg(["count", "mean"])
```
