```python
import pandas as pd

titanic = pd.read_csv("../data/titanic_propre.csv")

print(titanic.head())
```

```python
ports = pd.DataFrame({
    "embark_town": ["Southampton", "Cherbourg", "Queenstown"],
    "pays": ["Angleterre", "France", "Irlande"]
})

print(ports)
```

```python
titanic_pays = titanic.merge(ports, on="embark_town", how="left")

print(titanic_pays.head())
```

```python
print("Nombre de lignes avant :", len(titanic))
print("Nombre de lignes après :", len(titanic_pays))
```

```python
resultat = titanic_pays.groupby("pays").agg(
    nombre_passagers=("survived", "count"),
    taux_survie=("survived", "mean")
)

print(resultat)
```

```python
resultat["taux_survie"] = resultat["taux_survie"] * 100

print(resultat)
```

```python
ports_sans_queenstown = ports[
    ports["embark_town"] != "Queenstown"
]

print(ports_sans_queenstown)
```

```python
titanic_inner = titanic.merge(
    ports_sans_queenstown,
    on="embark_town",
    how="inner"
)

print(titanic_inner.head())
```

```python
titanic_left = titanic.merge(
    ports_sans_queenstown,
    on="embark_town",
    how="left"
)

print(titanic_left.head())
```

```python
print("Nombre de lignes Titanic :", len(titanic))
print("Jointure INNER :", len(titanic_inner))
print("Jointure LEFT :", len(titanic_left))
```

```python
print("Valeurs manquantes dans INNER :")
print(titanic_inner["pays"].isna().sum())

print("Valeurs manquantes dans LEFT :")
print(titanic_left["pays"].isna().sum())
```

```python
sans_pays = titanic_left[titanic_left["pays"].isna()]

print(sans_pays)
```

```python
print("Tableau Titanic :", titanic.shape)
print("Tableau avec pays :", titanic_pays.shape)
print("Jointure INNER :", titanic_inner.shape)
print("Jointure LEFT :", titanic_left.shape)
```
