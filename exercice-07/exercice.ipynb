```python
import seaborn as sns
import pandas as pd
titanic = sns.load_dataset("titanic") 
```

```python
nb_moins_18_ans=(titanic["age"]<18).sum()
print(nb_moins_18_ans) 
```

```python
billets_plus_cheres = titanic.nlargest(5, "fare")[["fare", "class", "embark_town"]]
billet_plus_cher = titanic["fare"].max()
print("Billet le plus cher :", round(billet_plus_cher, 2))
billets_plus_cheres
```

```python
premiere_classe = titanic[titanic["class"] == "First"]
part_femmes_1re = (premiere_classe["sex"] == "female").mean() * 100
print("Femmes en 1re classe :", round(part_femmes_1re, 1), "%")
```

```python
femmes_1re = titanic[(titanic["class"] == "First") & (titanic["sex"] == "female")]
survie_femmes_1re = femmes_1re["survived"].mean() * 100
print("Survie femmes 1re :", round(survie_femmes_1re, 1), "%")

hommes_3e = titanic[(titanic["class"] == "Third") & (titanic["sex"] == "male")]
survie_hommes_3e = hommes_3e["survived"].mean() * 100
print("Survie hommes 3e :", round(survie_hommes_3e, 1), "%")
```
