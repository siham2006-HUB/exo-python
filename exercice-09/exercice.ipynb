```python
import pandas as pd

df = pd.read_csv("../exercice-08/data/titanic_propre.csv")
df.shape
```

```python
survie_par_classe_et_sexe = df.groupby(["class", "sex"])["survived"].mean().unstack() * 100
survie_par_classe_et_sexe.round(1)
```

**Réponse 1 :** Oui, dans chaque classe les femmes ont beaucoup mieux survécu que les hommes : 96,8 % contre 36,9 % en 1re, 92,1 % contre 15,7 % en 2e et 50 % contre 13,5 % en 3e.

```python
df["tranche_age"] = pd.cut(df["age"], bins=[0, 12, 18, 60, 100],
                           labels=["Enfant (0-12)", "Ado (12-18)", "Adulte (18-60)", "Senior (60+)"])
survie_par_tranche_age = df.groupby("tranche_age", observed=True)["survived"].mean() * 100
survie_par_tranche_age.round(1)
```

**Réponse 2 :** Oui, les enfants ont été en partie prioritaires : 58 % des moins de 12 ans ont survécu, contre 36,6 % des adultes et 22,7 % des plus de 60 ans. Le taux de survie baisse avec l'âge.

```python
prix_moyen_par_port = df.groupby("embark_town")["fare"].mean().sort_values(ascending=False)
print(prix_moyen_par_port.round(2))

repartition_classes_par_port = pd.crosstab(df["embark_town"], df["class"], normalize="index") * 100
repartition_classes_par_port.round(1)
```

```python
survie_seul_ou_accompagne = df.groupby("alone")["survived"].mean() * 100
survie_seul_ou_accompagne.round(1)
```

**Réponse 3 :** Cherbourg a le prix moyen le plus élevé (59,95 £, contre 27,24 £ à Southampton et 13,28 £ à Queenstown). La répartition des classes l'explique : la moitié des passagers de Cherbourg (50,6 %) voyageaient en 1re classe, alors qu'à Queenstown 93,5 % étaient en 3e.

**Réponse 4 :** Voyager seul diminuait les chances de survie : 30,4 % des passagers seuls ont survécu, contre 50,6 % de ceux qui voyageaient en famille.
