## Constats

1. **Données incomplètes** : `deck` est vide à 77 % (688/891) et `age` manque pour 177 passagers (20 %).
2. **La 3e classe est majoritaire** : 491 passagers sur 891 (55 %).
3. **Southampton est le port principal** : 644 passagers (72 %). Queenstown n'a presque que des passagers de 3e classe (72 sur 77).


## Types mal choisis ou redondants

- `survived` (0/1) et `alive` (yes/no) donnent la même information.
- `pclass` (1/2/3) et `class` (First/Second/Third) sont redondantes ; `pclass` est un entier alors que c'est une catégorie.
- `embarked` (S/C/Q) et `embark_town` sont redondantes.
- `survived` devrait être un booléen, `sex` et `embarked` des catégories plutôt que du texte.
